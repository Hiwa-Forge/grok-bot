# Grok bot — xAI

Ask Grok a question with web search, X search and code execution enabled.

**Setup:**

```bash
export XAI_API_KEY="your-key-here"
```

Or copy `.env.example` to `.env` and fill it in. `.env` is ignored by git.

````python
import os

from xai_sdk import Client
from xai_sdk.chat import user
from xai_sdk.tools import web_search, x_search, code_execution

api_key = os.getenv("XAI_API_KEY")
if not api_key:
    raise ValueError("XAI_API_KEY is not set. Export it before running this notebook.")

client = Client(api_key=api_key)
````

````python
chat = client.chat.create(
    model="grok-4-1-fast-reasoning",
    tools=[
        web_search(),
        x_search(),
        code_execution(),
    ],
)
````

````python
chat.append(user("What's the latest news on xAI?"))
response = chat.sample()
print(response.content)
````
